```python
# load dataset
df = pd.read_csv('BankChurners.csv')

# First look at the data
print("Shape:", df.shape)
print("\nFirst 5 rows:")
df.head()



```

```python
3 # Explore the dataset
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nBasic statistics:")
df.describe()
```

```python
# Drop irrelevant columns 
df = df.drop(columns=[col for col in df.columns
                      if 'Naive_Bayes' in col])
# Confirm Columns dropped
print("Remaining columns:", df.shape[1])

# CHeck churn distribution
print("\nChurn Distribution:")
print(df['Attrition_Flag'].value_counts())
print("\nChurn Percentage:")
print(df['Attrition_Flag'].value_counts(normalize=True) *100)
```

```python
# ANALYSIS 1 - Churn by Gender
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Gender', hue='Attrition_Flag')
plt.title('Churn by Gender')
plt.xlabel('Gender') 
plt.ylabel('Number of Customers')
plt.legend(title='Status') 
plt.tight_layout()
plt.show()
```

```python
# ANALYSIS 2 - Churn by Income Category
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Income_Category', hue='Attrition_Flag')
plt.title('Churn by Income Category')
plt.xlabel('Income Category')
plt.ylabel('Number of Customers')
plt.legend(title='Status')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# ANALYSIS 3 - Churn by Card Category
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Card_Category', hue='Attrition_Flag')
plt.title('Churn by Card Category')
plt.xlabel('Card Category')
plt.ylabel('Number of Customers')
plt.legend(title='Status')
plt.tight_layout()
plt.show()
```

```python
# ANALYSIS 4 - Age distribution of churned vs existing
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Customer_Age',hue='Attrition_Flag', bins=20)
plt.title('Age Distribution - Churned vs Existing')
plt.xlabel('Customer Age')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# ANALYSIS 5 -Transaction Amount by Churn Status
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Attrition_Flag', y='Total_Trans_Amt')
plt.title('Total Transaction Amount - Churned vs Existing')
plt.xlabel('Customer Status')
plt.ylabel('Total Transaction Amount')
plt.tight_layout()
plt.show()

# ANALYSIS 6 - Months Inactive by Churn Status
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Attrition_Flag', y='Months_Inactive_12_mon')
plt.title('Monts Inactive - Churned vs Existing')
plt.xlabel('Customer Status')
plt.ylabel('Months Inactive (Last 12 months)')
plt.tight_layout()
plt.show()
```

```python
# ANALYSIS 7 - Correlation Heatmap
plt.figure(figsize=(12,8))
numeric_df = df.select_dtypes(include=['int64','float64'])
sns.heatmap(numeric_df.corr(), annot=True, fmt='2f', cmap='coolwarm')
plt.title('Correlation Heatmap of Numeric Features')
plt.tight_layout()
plt.show()
```
